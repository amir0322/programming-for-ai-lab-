Implement the N-Queens Problem (Dynamic)

```python
def print_solution(matrix):
    for board_row in matrix:
        row_text = " ".join("Q" if slot == 1 else "." for slot in board_row)
        print(row_text)
    print()


def is_placement_safe(matrix, r_idx, c_idx):
    total_rows = len(matrix)

    
    for checked_row in range(r_idx):
        if matrix[checked_row][c_idx] == 1:
            return False

        offset = r_idx - checked_row
        left_col = c_idx - offset
        right_col = c_idx + offset

        left_blocked = left_col >= 0 and matrix[checked_row][left_col] == 1
        right_blocked = right_col < total_rows and matrix[checked_row][right_col] == 1
        if left_blocked or right_blocked:
            return False

    return True


def enumerate_nqueen_solutions(matrix, row_pointer=0):
    board_size = len(matrix)

    if row_pointer == board_size:
        print_solution(matrix)
        return True

    solution_exists = False
    for trial_col in range(board_size):
        if not is_placement_safe(matrix, row_pointer, trial_col):
            continue

        matrix[row_pointer][trial_col] = 1
        has_solution_below = enumerate_nqueen_solutions(matrix, row_pointer + 1)
        solution_exists = solution_exists or has_solution_below
        matrix[row_pointer][trial_col] = 0

    return solution_exists


dimension = 4
queen_board = [[0] * dimension for _ in range(dimension)]
enumerate_nqueen_solutions(queen_board)
```
