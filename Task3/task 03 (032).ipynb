Task 1: Solve the Water Jug Problem using Depth-First Search (DFS)

and display the rule/action applied at each transition.

```python
MAX_A = 4
MAX_B = 3
GOAL = 2

visited = set()
path = []

def next_states(state):
    a, b = state
    steps = []

    if a < MAX_A:
        steps.append(((MAX_A, b), "fill A"))
    if b < MAX_B:
        steps.append(((a, MAX_B), "fill B"))

    if a > 0:
        steps.append(((0, b), "empty A"))
    if b > 0:
        steps.append(((a, 0), "empty B"))

    if a > 0 and b < MAX_B:
        transfer = min(a, MAX_B - b)
        steps.append(((a - transfer, b + transfer), f"pour {transfer} from A->B"))
    if b > 0 and a < MAX_A:
        transfer = min(b, MAX_A - a)
        steps.append(((a + transfer, b - transfer), f"pour {transfer} from B->A"))

    return steps


def dfs(current):
    a, b = current
    visited.add(current)
    path.append(current)

    if a == GOAL or b == GOAL:
        print("Reached goal at", current)
        return True

    for new_state, action in next_states(current):
        if new_state in visited:
            continue
        print(f"=> {action}, state: {new_state}")
        if dfs(new_state):
            return True

    path.pop()
    return False

# start search
start_state = (0, 0)
if dfs(start_state):
    print("\nPath:")
    for s in path:
        print(s)
else:
    print("No solution found")
```
